```python
import kagglehub
path = kagglehub.dataset_download("ramjidoolla/ipl-data-set")
```

```python
import pandas as pd

df = pd.read_csv('/kaggle/input/ipl-data-set/matches.csv')

df.head(15)
```

```python
df.info
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna()
```

```python
print(df.shape)
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt

df['Season'].value_counts().sort_index().plot(
    kind='bar',
    figsize=(8,5)
)

plt.title("Matches Played Per Season")
plt.xlabel("Season")
plt.ylabel("Number of Matches")
plt.show()
```

```python
df['winner'].value_counts().head(10).plot(
    kind='bar',
    figsize=(8,5)
)

plt.title("Top Winning IPL Teams")
plt.xlabel("Team")
plt.ylabel("Wins")
plt.show()
```

```python
df['toss_decision'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title("Toss Decisions")
plt.ylabel("")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Remove matches won by wickets (win_by_runs = 0)
runs_margin = df[df['win_by_runs'] > 0]

# Top 10 most common winning margins
top_runs = runs_margin['win_by_runs'].value_counts().head(10)

plt.figure(figsize=(10,6))
top_runs.plot(kind='bar')

plt.title("Top 10 Most Common Winning Margins by Runs in IPL")
plt.xlabel("Winning Margin (Runs)")
plt.ylabel("Number of Matches")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

wicket_wins = df[df['win_by_wickets'] > 0]['win_by_wickets']

wicket_wins.value_counts().sort_index().plot(
    kind='bar',
    figsize=(8,5)
)

plt.title("Winning Margins by Wickets")
plt.xlabel("Number of Wickets")
plt.ylabel("Number of Matches")

plt.show()
```

```python
import matplotlib.pyplot as plt

top_venues = df['venue'].value_counts().head(10)

plt.figure(figsize=(10,6))

top_venues.plot(
    kind='bar',
    edgecolor='black'
)

plt.title('Top 10 IPL Venues by Number of Matches', fontsize=14)
plt.xlabel('Venue', fontsize=12)
plt.ylabel('Number of Matches', fontsize=12)

plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

top_winners = df['winner'].value_counts().head(10)

plt.figure(figsize=(10,6))

top_winners.plot(
    kind='bar',
    edgecolor='black'
)

plt.title('Top 10 IPL Teams by Total Match Wins', fontsize=14)
plt.xlabel('Team', fontsize=12)
plt.ylabel('Number of Wins', fontsize=12)

plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```
